## 6. Limpeza e preparação

A partir daqui o `df` deixa de ser a base original e passa a ser a base tratada.

A função `tratar` (arquivo `scripts/tratar_base.py`, detalhes em `tratamento_dados.md`) faz a primeira parte da limpeza:

- unifica `DIA`, `MES`, `ANO` e `HORA` em `DATA E HORA`;
- substitui `LOG1`, `LOG2` e `NUMERO` por `BAIRRO` e `RUA`, obtidas das coordenadas;
- mantém `COD_ACIDENTE` como texto.

```python
import sys
import os

# Caminho real da sua pasta de scripts no Google Drive
caminho_real_scripts = '/content/drive/MyDrive/analise_sinistros_transito_amc/Análises/0. Scripts'

# Criamos um link simbólico chamado 'scripts' apontando para '0. Scripts'
# Isso resolve a importação interna 'from scripts.localizacao ...' de forma limpa!
link_atalho = '/content/scripts'

if os.path.exists(caminho_real_scripts):
    # Se o atalho antigo existir ou for um link quebrado, removemos para recriar com segurança
    if os.path.islink(link_atalho) or os.path.exists(link_atalho):
        try:
            os.unlink(link_atalho)
        except Exception:
            pass

    # Cria o link simbólico do sistema operacional
    os.symlink(caminho_real_scripts, link_atalho)

    # Adiciona o diretório atual '/content' ao sys.path para que o Python enxergue o atalho 'scripts'
    if '/content' not in sys.path:
        sys.path.append('/content')

    print("Atalho simbólico de '0. Scripts' para 'scripts' criado com sucesso!")
else:
    print(f"AVISO: A pasta original não foi encontrada no Drive: {caminho_real_scripts}")

# Agora a importação funcionará perfeitamente, mapeando todas as referências internas!
from scripts.tratar_base import tratar
```

### Texto

Padronização das categorias: sem espaços sobrando e com só a primeira letra maiúscula, para unificar grafias como `Meio de Quadra`, `Meio De Quadra` e `Meio de quadra`.

```python
categoricas = ["TIPOCRUZAMENTO", "CONTROLETRAFEGO", "USOSOLO", "NATUREZA", "SEVERIDADE"]

categorias_antes = df[categoricas].nunique()

for coluna in categoricas:
    df[coluna] = (
        df[coluna]
        .str.strip()
        .str.replace(r"\s+", " ", regex=True)
        .str.capitalize()
        .replace({"Duplo t": "Duplo T"})
    )

pd.DataFrame({"categorias_antes": categorias_antes, "categorias_depois": df[categoricas].nunique()})
```

### Tipos

```python
# Primeiro, executamos o tratamento dos dados utilizando a função importada para criar 'DATA E HORA' e 'BAIRRO'
df = tratar(caminho_drive)

# Agora as conversões de tipos e formatações adicionais funcionarão corretamente
df["COD_ACIDENTE"] = df["COD_ACIDENTE"].astype("string")
df["DATA E HORA"] = pd.to_datetime(df["DATA E HORA"])

df[categoricas + ["BAIRRO"]] = df[categoricas + ["BAIRRO"]].astype("category")
df["SEVERIDADE"] = df["SEVERIDADE"].cat.reorder_categories(
    ["Ileso", "Ferido", "Fatal", "Não informado"]
)

df.dtypes
```

### Novas variáveis

| Coluna | Como é calculada |
|---|---|
| `DIA_SEMANA` | Dia da semana de `DATA E HORA` (Segunda a Domingo). |
| `FIM_DE_SEMANA` | `True` para sábado e domingo. |
| `FAIXA_HORARIA` | Madrugada (0h–5h59), Manhã (6h–11h59), Tarde (12h–17h59), Noite (18h–23h59). |
| `TOTAL_VITIMAS` | `MORTOS` + `FERIDOS`. |
| `TOTAL_VEICULOS` | Soma das nove colunas de veículos. |

```python
dias = ["Segunda", "Terça", "Quarta", "Quinta", "Sexta", "Sábado", "Domingo"]
dia = df["DATA E HORA"].dt.dayofweek

df["DIA_SEMANA"] = pd.Categorical.from_codes(dia, categories=dias, ordered=True)
df["FIM_DE_SEMANA"] = dia >= 5
df["FAIXA_HORARIA"] = pd.cut(
    df["DATA E HORA"].dt.hour,
    bins=[0, 6, 12, 18, 24],
    labels=["Madrugada", "Manhã", "Tarde", "Noite"],
    right=False,
)

veiculos = [
    "AUTOMOVEL", "ONIBUS", "BICICLETA", "CAMINHAO", "MICROONIBUS",
    "CICLOMOTOR", "MOTOCICLETA", "TRACAOANIMAL", "TREM",
]
df["TOTAL_VITIMAS"] = df["MORTOS"] + df["FERIDOS"]
df["TOTAL_VEICULOS"] = df[veiculos].sum(axis=1)

df[["DATA E HORA", "DIA_SEMANA", "FIM_DE_SEMANA", "FAIXA_HORARIA", "TOTAL_VITIMAS", "TOTAL_VEICULOS"]].head()
```

## 7. Dicionário de Dados (depois da limpeza)

Base preparada: `dataset/1_sinistros_transito_2015_2024_preparada.parquet`: 130.823 linhas, 28 colunas.
Cada linha é um sinistro de trânsito registrado em Fortaleza entre 2015 e 2024.

### Identificação, tempo e local

| Coluna | Tipo | Classificação | Descrição | Valores |
|---|---|---|---|---|
| `COD_ACIDENTE` | texto | Identificador | Código único do sinistro. Numérico até 2020 e alfanumérico a partir de 2021. | 130.820 códigos distintos; 3 vazios |
| `DATA E HORA` | data e hora | Quantitativa contínua (temporal) | Momento do sinistro. Criada a partir de `DIA`, `MES`, `ANO` e `HORA`. | 01/01/2015 00:11 a 31/12/2024 23:07 |
| `BAIRRO` | categoria | Qualitativa nominal | Bairro de Fortaleza onde o ponto está. Derivado de `LATITUDE` e `LONGITUDE`. | 121 bairros; 74 vazios (pontos fora do município) |
| `RUA` | texto | Qualitativa nominal | Via mais próxima do ponto. Derivada de `LATITUDE` e `LONGITUDE`. | 4.127 vias distintas |
| `LATITUDE` | decimal | Quantitativa contínua | Latitude do sinistro, em graus decimais. | -3,896 a -3,693 |
| `LONGITUDE` | decimal | Quantitativa contínua | Longitude do sinistro, em graus decimais. | -38,638 a -38,410 |

### Características do sinistro

| Coluna | Tipo | Classificação | Descrição | Valores |
|---|---|---|---|---|
| `TIPOCRUZAMENTO` | categoria | Qualitativa nominal | Configuração da via no local. | 10 categorias (ex.: Meio de quadra, Cruz, T, Rotatória); 12,46% "Não informado" |
| `CONTROLETRAFEGO` | categoria | Qualitativa nominal | Controle de tráfego existente no local. | 13 categorias (ex.: Pare, Semáforo, Não há controle); 59,91% "Não informado" |
| `USOSOLO` | categoria | Qualitativa nominal | Posição do sinistro em relação à via. | Longo da via, Cruzamento, Não informado (21,80%) |
| `NATUREZA` | categoria | Qualitativa nominal | Tipo de sinistro. | 20 categorias (ex.: Colisão, Queda, Atropelamento); 0,96% "Não informado" |
| `SEVERIDADE` | categoria | Qualitativa ordinal | Gravidade do sinistro. | Ileso, Ferido, Fatal, Não informado (7,06%) |

### Pessoas envolvidas

| Coluna | Tipo | Classificação | Descrição | Valores |
